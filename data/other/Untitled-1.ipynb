```python
DATABASE_PATH = "../../database/hockey_v16_test.db"
```

```python
db_session = GetDatabaseSession(db_path=DATABASE_PATH)
db_session.set_up_connection()
query_o = dq.ParsedDBQuery(db_session=db_session.session)
```

```python
query = text("""
    SELECT DISTINCT players.uid
    FROM player_stats
    JOIN players ON players.id = player_stats.player_id
    JOIN players_draft ON players.id = players_draft.player_id
    JOIN leagues ON player_stats.league_id = leagues.id
    JOIN seasons ON player_stats.season_id = seasons.id
    JOIN teams ON teams.id = player_stats.team_id
    LEFT JOIN players_nationalities ON players_nationalities.player_id = player_stats.player_id
    WHERE nationality_id IS NULL
""")

result = db_session.session.execute(query).fetchall()
```

```python
list_uids = []
for tuple_ in result:
    list_uids.append(tuple_[0])
```

```python
import json
with open('nationality_uid.json', 'w') as f:
    json.dump(list_uids, f)
```
